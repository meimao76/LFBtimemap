```python
import json

# 输入和输出的文件名
input_file = 'incident_standardized.geojson'
output_file = 'incident_standardized_filtered.geojson'

# 读取原始 GeoJSON 文件
with open(input_file, 'r', encoding='utf-8') as f:
    data = json.load(f)

# 过滤掉 geometry 坐标为 [0, 0] 的 feature
filtered_features = [
    feature for feature in data['features']
    if feature.get('geometry', {}).get('coordinates') != [0, 0]
]

# 更新 features
data['features'] = filtered_features

# 将过滤后的数据写入新的 GeoJSON 文件
with open(output_file, 'w', encoding='utf-8') as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print(f"过滤后的数据已保存到 {output_file}")
```
